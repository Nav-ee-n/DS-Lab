```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
data,_=make_blobs(n_samples=300)
print(data)
```

```python
kmeans=KMeans(n_clusters=5,random_state=42,n_init='auto').fit(data)
```

```python
labels=kmeans.labels_
labels
```

```python
set(labels)
```

```python
centroids=kmeans.cluster_centers_
#stores the coordinates of the centroids for these five clusters
centroids
```

```python
import matplotlib.pyplot as plt
plt.scatter(data[:,0],data[:,1],c=labels,cmap='viridis') #this will plot the data points and colouring each point according to its cluster labels
plt.scatter(centroids[:,0],centroids[:,1],marker='X',color='red')
plt.legend(list(labels))
plt.show()
```

```python
kmeans.predict([[2,-4],[6,6],[6,-2]])
```
